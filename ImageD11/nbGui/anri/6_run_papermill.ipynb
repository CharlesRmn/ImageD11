```python
import ImageD11.sinograms.dataset
from ImageD11.nbGui import segmenter_gui
from ImageD11.nbGui import nb_utils as utils
from esrf_pathlib import ESRFPath as Path

import pprint
```

```python
ds = ImageD11.sinograms.dataset.load(dset_path)
print(ds)
```

```python
# try to get list of samples automatically
if sample_list is None:
    sample_list = sorted([name for name in os.listdir(ds.dataroot) if os.path.isdir(os.path.join(ds.dataroot, name))])

print(sample_list)
```

```python
if samples_dict is None:
    samples_dict = utils.find_datasets_to_process(ds.dataroot, skips_dict, dset_prefix, sample_list)

print(samples_dict)
```

```python
extensions = {'.ipynb'}#{'.h5', '.xdmf', '.ctf', '.ipynb', '*.bak', '*.edf'}
do_del = {'tomo_2_map','tomo_3_refinement','4_visualise'}

for i in samples_dict['pureIron_2_heating_3']:
    dset = Path.cwd().replace_fields(dataset=i)
    for file in dset.iterdir():
        # print(f'checking {file.name}')
        if file.is_file() and file.suffix in extensions and any(pattern in file.name for pattern in do_del):
            print(f"Deleting {file}")
            file.unlink()
```
